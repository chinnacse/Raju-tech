```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read Image
image = cv2.imread("images (1).jpg")

if image is None:
    print("Image not found!")
else:

    # Convert BGR to RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Average Filter
    average = cv2.blur(image, (5,5))

    # Gaussian Filter
    gaussian = cv2.GaussianBlur(image, (5,5), 0)

    # Median Filter
    median = cv2.medianBlur(image, 5)

    # Sharpen Filter
    kernel = np.array([[0,-1,0],
                       [-1,5,-1],
                       [0,-1,0]])

    sharpen = cv2.filter2D(image, -1, kernel)

    # Display Results
    plt.figure(figsize=(15,10))

    plt.subplot(2,3,1)
    plt.imshow(image)
    plt.title("Original")
    plt.axis("off")

    plt.subplot(2,3,2)
    plt.imshow(average)
    plt.title("Average Filter")
    plt.axis("off")

    plt.subplot(2,3,3)
    plt.imshow(gaussian)
    plt.title("Gaussian Filter")
    plt.axis("off")

    plt.subplot(2,3,4)
    plt.imshow(median)
    plt.title("Median Filter")
    plt.axis("off")

    plt.subplot(2,3,5)
    plt.imshow(sharpen)
    plt.title("Sharpen Filter")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
