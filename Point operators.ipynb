```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("348106.jpg")

if image is None:
    print("Image not found!")
else:
    # Convert BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Increase Brightness
    brightness_increased = cv2.convertScaleAbs(image, alpha=1.0, beta=60)

    brightness_rgb = cv2.cvtColor(brightness_increased, cv2.COLOR_BGR2RGB)

    # Display
    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.imshow(brightness_rgb)
    plt.title("Brightness Increased")
    plt.axis("off")

    plt.show()
```
