```python
import cv2
import numpy as np

# Read image in grayscale
image = cv2.imread("VK18.jpg", cv2.IMREAD_GRAYSCALE)

print(image.shape)
```

```python
pixel = image[100, 150] #Get the Pixel Value of One Location
print(pixel)
```

```python
print(image[100:105,100:105]) 
#Meaning:

#Row = 100
#Column = 150
#Brightness = 132

#Since it is an 8-bit grayscale image, pixel values range from:
#0   → Black
#255 → White
#128 → Gray
```

```python
blur = cv2.GaussianBlur(image, (9,9), 0)

print(blur[100:105,100:105])
#0   → Black
#255 → White
#128 → Gray
```

```python
noise = np.random.normal(0,20,image.shape) # Create noisy image

noisy = image + noise
noisy = np.clip(noisy,0,255).astype(np.uint8)

print(noisy[100:105,100:105])
```

```python
print("Original Pixel :", image[100,100]) #compare both values
print("Blur Pixel     :", blur[100,100])
print("Noisy Pixel    :", noisy[100,100])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(image, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(blur, cmap='gray')
plt.title("Blurred")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(noisy, cmap='gray')
plt.title("Noisy")
plt.axis("off")

plt.show()
```
