```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("VK18.jpg")
if image is None:
    print("Image not found!")
else:
    print("Image loaded successfully!")
```

```python
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6,8))
plt.imshow(rgb)
plt.title("Original RGB Image")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(6,8))
plt.imshow(gray, cmap='gray')
plt.title("Grayscale")
plt.axis("off")
plt.show()
```

```python
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(hsv[:,:,0], cmap='hsv')
plt.title("Hue")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(hsv[:,:,1], cmap='gray')
plt.title("Saturation")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(hsv[:,:,2], cmap='gray')
plt.title("Value")
plt.axis("off")

plt.show()
```

```python
lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(lab[:,:,0], cmap='gray')
plt.title("L (Lightness)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(lab[:,:,1], cmap='gray')
plt.title("A Channel")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(lab[:,:,2], cmap='gray')
plt.title("B Channel")
plt.axis("off")

plt.show()
```

```python
ycrcb = cv2.cvtColor(image, cv2.COLOR_BGR2YCrCb)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(ycrcb[:,:,0], cmap='gray')
plt.title("Y (Luminance)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(ycrcb[:,:,1], cmap='gray')
plt.title("Cr")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(ycrcb[:,:,2], cmap='gray')
plt.title("Cb")
plt.axis("off")

plt.show()
```

```python
hls = cv2.cvtColor(image, cv2.COLOR_BGR2HLS)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(hls[:,:,0], cmap='hsv')
plt.title("Hue")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(hls[:,:,1], cmap='gray')
plt.title("Lightness")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(hls[:,:,2], cmap='gray')
plt.title("Saturation")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(14,10))

plt.subplot(2,3,1)
plt.imshow(rgb)
plt.title("RGB")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB))
plt.title("HSV")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(cv2.cvtColor(lab, cv2.COLOR_LAB2RGB))
plt.title("LAB")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(cv2.cvtColor(ycrcb, cv2.COLOR_YCrCb2RGB))
plt.title("YCrCb")
plt.axis("off")

plt.subplot(2,3,6)
plt.imshow(cv2.cvtColor(hls, cv2.COLOR_HLS2RGB))
plt.title("HLS")
plt.axis("off")

plt.tight_layout()
plt.show()
```
