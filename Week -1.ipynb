```python
import cv2

# Open the video file
video = cv2.VideoCapture(r"C:\Users\mchin\Downloads\WhatsApp Video 2026-08-04 at 14.51.38.mp4")

# Check whether the video opened successfully
if not video.isOpened():
    print("Error: Cannot open the video.")
    exit()

print("Video opened successfully.")

# Read the video frame by frame
while True:

    # Read one frame
    success, frame = video.read()

    # Stop when the video ends
    if not success:
        print("End of Video")
        break

    # Display the frame
    cv2.imshow("Video Player", frame)

    # Press Q to quit early
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

# Release resources
video.release()

# Close all windows
cv2.destroyAllWindows()
```
