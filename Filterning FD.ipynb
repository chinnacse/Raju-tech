```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read color image
img = cv2.imread("")

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# ---------------------------------------------------
# Function for Frequency Domain Filtering
# ---------------------------------------------------

def frequency_filter(channel, radius=50, filter_type="low"):

    # Fourier Transform
    f = np.fft.fft2(channel)

    # Move low frequencies to center
    fshift = np.fft.fftshift(f)

    # Get image dimensions
    rows, cols = channel.shape
    crow, ccol = rows // 2, cols // 2

    # Create filter mask
    mask = np.zeros((rows, cols), np.float32)

    if filter_type == "low":
        # Low-Pass Filter
        cv2.circle(mask, (ccol, crow), radius, 1, -1)

    elif filter_type == "high":
        # High-Pass Filter
        mask[:] = 1
        cv2.circle(mask, (ccol, crow), radius, 0, -1)

    # Apply filter
    filtered = fshift * mask

    # Inverse Fourier Transform
    f_ishift = np.fft.ifftshift(filtered)
    result = np.fft.ifft2(f_ishift)

    # Convert to image
    result = np.abs(result)

    return result


# ---------------------------------------------------
# Split RGB Channels
# ---------------------------------------------------

R, G, B = cv2.split(img_rgb)

# ---------------------------------------------------
# Apply Low-Pass Filter
# ---------------------------------------------------

R_low = frequency_filter(R, 50, "low")
G_low = frequency_filter(G, 50, "low")
B_low = frequency_filter(B, 50, "low")

low_pass_color = cv2.merge([
    R_low.astype(np.uint8),
    G_low.astype(np.uint8),
    B_low.astype(np.uint8)
])


# ---------------------------------------------------
# Apply High-Pass Filter
# ---------------------------------------------------

R_high = frequency_filter(R, 50, "high")
G_high = frequency_filter(G, 50, "high")
B_high = frequency_filter(B, 50, "high")

high_pass_color = cv2.merge([
    R_high.astype(np.uint8),
    G_high.astype(np.uint8),
    B_high.astype(np.uint8)
])


# ---------------------------------------------------
# Display Results
# ---------------------------------------------------

plt.figure(figsize=(15, 10))

# Original
plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Color Image")
plt.axis("off")

# Low Pass
plt.subplot(2, 3, 2)
plt.imshow(low_pass_color)
plt.title("Low-Pass Filter")
plt.axis("off")

# High Pass
plt.subplot(2, 3, 3)
plt.imshow(high_pass_color)
plt.title("High-Pass Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```
